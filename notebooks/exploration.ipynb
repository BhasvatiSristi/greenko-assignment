Part B — Dataset Understanding

For Part B, I will be using the 90-day turbine telemetry dataset (turbine_90) and the 90-day DAM price dataset (dam_price_90).

turbine_90 contains 10,725 rows and 5 columns. With 5 turbines recorded hourly over 90 days, we would expect 90 × 24 × 5 = 10,800 rows. Therefore, the dataset has 75 missing telemetry records.
dam_price_90 contains 2,160 rows and 2 columns, which matches the expected 90 × 24 = 2,160 hourly observations.
Both datasets contain no null values and no duplicate rows.

The missing telemetry records in turbine_90 therefore correspond to missing rows rather than null values, which is an important data-quality issue that needs to be investigated during the EDA. The provided README also explicitly states that missing telemetry rows should not be assumed to represent zero generation.
